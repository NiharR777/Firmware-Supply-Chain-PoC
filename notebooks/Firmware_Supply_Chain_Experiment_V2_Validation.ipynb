```python
# ================================================================
# CELL 2 — EXPERIMENT V2 PROTOCOL CONFIGURATION
# ================================================================

from pathlib import Path
import json
import hashlib
from datetime import datetime, timezone

ROOT = Path("/content/Firmware-Supply-Chain-PoC")

DATASET_DIR = ROOT / "data" / "synthetic" / "supply_chain_poc" / "dataset"
V2_DIR = ROOT / "artifacts" / "ml_experiment_v2"

V2_DIR.mkdir(parents=True, exist_ok=True)

# ----------------------------------------------------------------
# Existing experiment — reference only
# ----------------------------------------------------------------

OLD_SEED = 20260828
OLD_RECORD_COUNT = 5000
OLD_LINEAGE_COUNT = 500
OLD_RECORDS_PER_LINEAGE = 10
OLD_GENERATOR_VERSION = "2.0.0"

# ----------------------------------------------------------------
# V2 independent holdout design
# ----------------------------------------------------------------

V2_SEED = 20261003
V2_RECORD_COUNT = 1000
V2_LINEAGE_COUNT = 100
V2_RECORDS_PER_LINEAGE = 10

V2_GENERATOR_VERSION = OLD_GENERATOR_VERSION

V2_RUN_ID = "V2-HOLDOUT-20261003"

# The V2 holdout must remain completely separate from the old
# 5,000-record experiment.
V2_OUTPUT_DIR = V2_DIR / "holdout"

PROTOCOL = {
    "experiment": "Firmware Supply-Chain PoC — Experiment V2",
    "purpose": "Independent holdout preparation",
    "status": "PROTOCOL_ONLY",
    "old_experiment": {
        "seed": OLD_SEED,
        "record_count": OLD_RECORD_COUNT,
        "lineage_count": OLD_LINEAGE_COUNT,
        "records_per_lineage": OLD_RECORDS_PER_LINEAGE,
        "generator_version": OLD_GENERATOR_VERSION,
    },
    "v2_holdout": {
        "seed": V2_SEED,
        "record_count": V2_RECORD_COUNT,
        "lineage_count": V2_LINEAGE_COUNT,
        "records_per_lineage": V2_RECORDS_PER_LINEAGE,
        "generator_version": V2_GENERATOR_VERSION,
        "run_id": V2_RUN_ID,
    },
    "separation_rule": (
        "The V2 holdout must not reuse records, record IDs, "
        "lineages, or opened test observations from Experiment V1."
    ),
    "training_rule": (
        "No V2 model training or model selection is permitted "
        "in this protocol stage."
    ),
    "synthetic_data_only": True,
    "created_at_utc": datetime.now(timezone.utc).isoformat(),
}

protocol_path = V2_DIR / "v2_protocol.json"
protocol_path.write_text(
    json.dumps(PROTOCOL, indent=2, sort_keys=True) + "\n",
    encoding="utf-8",
)

print("=" * 70)
print("EXPERIMENT V2 — PROTOCOL CONFIGURATION")
print("=" * 70)

print("V2 seed:", V2_SEED)
print("V2 records:", V2_RECORD_COUNT)
print("V2 lineages:", V2_LINEAGE_COUNT)
print("Records per lineage:", V2_RECORDS_PER_LINEAGE)
print("V2 run ID:", V2_RUN_ID)
print("Output directory:", V2_OUTPUT_DIR)
print("Training enabled:", False)
print("Protocol saved:", protocol_path.exists())

print("=" * 70)
```

```python
# ================================================================
# CELL 3 — V1 / V2 ISOLATION CHECK
# ================================================================

import pandas as pd

split_path = (
    ROOT
    / "data"
    / "synthetic"
    / "supply_chain_poc"
    / "dataset"
    / "split_assignments.csv"
)

old_splits = pd.read_csv(split_path)

old_record_ids = set(old_splits["record_id"].astype(str))
old_package_ids = set(old_splits["package_id"].astype(str))
old_group_ids = set(old_splits["group_id"].astype(str))

# V2 uses a completely separate ID namespace.
v2_record_ids = {
    f"V2-REC-{i:05d}"
    for i in range(1, V2_RECORD_COUNT + 1)
}

v2_package_ids = {
    f"V2-PKG-{i:04d}"
    for i in range(1, V2_LINEAGE_COUNT + 1)
}

v2_lineage_ids = {
    f"V2-LINEAGE-{i:04d}"
    for i in range(1, V2_LINEAGE_COUNT + 1)
}

print("=" * 70)
print("V1 / V2 ISOLATION CHECK")
print("=" * 70)

print("V1 records:", len(old_record_ids))
print("V1 packages:", len(old_package_ids))
print("V1 groups:", len(old_group_ids))

print("Proposed V2 records:", len(v2_record_ids))
print("Proposed V2 packages:", len(v2_package_ids))
print("Proposed V2 lineages:", len(v2_lineage_ids))

record_overlap = old_record_ids.intersection(v2_record_ids)
package_overlap = old_package_ids.intersection(v2_package_ids)
lineage_overlap = old_group_ids.intersection(v2_lineage_ids)

print("\nRecord ID overlap:", len(record_overlap))
print("Package ID overlap:", len(package_overlap))
print("Lineage/group overlap:", len(lineage_overlap))

assert not record_overlap, f"Record collision: {record_overlap}"
assert not package_overlap, f"Package collision: {package_overlap}"
assert not lineage_overlap, f"Lineage collision: {lineage_overlap}"

print("\nV1 / V2 identifier isolation: PASS")
print("=" * 70)
```

```python
# ================================================================
# CELL 4 — GENERATE INDEPENDENT V2 HOLDOUT
# ================================================================

import csv
import json
import random
import hashlib
from pathlib import Path
from datetime import datetime, timezone

# ------------------------------------------------
# Output paths
# ------------------------------------------------

HOLDOUT_DIR = V2_OUTPUT_DIR

OBSERVED_DIR = HOLDOUT_DIR / "observations"
REFERENCE_DIR = HOLDOUT_DIR / "reference"
LABEL_DIR = HOLDOUT_DIR / "labels"

for directory in [HOLDOUT_DIR, OBSERVED_DIR, REFERENCE_DIR, LABEL_DIR]:
    directory.mkdir(parents=True, exist_ok=True)


# ------------------------------------------------
# Deterministic V2 random generator
# ------------------------------------------------

rng = random.Random(V2_SEED)


# ------------------------------------------------
# V2 scenario distribution
# ------------------------------------------------

SCENARIOS = [
    "valid",
    "benign_exception",
    "vulnerable_component",
    "hash_mismatch",
    "invalid_signature",
    "missing_sbom",
    "unexpected_component",
    "untrusted_vendor",
    "rollback",
    "stale_evidence",
]

scenario_per_type = V2_RECORD_COUNT // len(SCENARIOS)

scenario_assignments = []

for scenario in SCENARIOS:
    scenario_assignments.extend(
        [scenario] * scenario_per_type
    )

rng.shuffle(scenario_assignments)


# ------------------------------------------------
# Synthetic V2 records
# ------------------------------------------------

records = []

for i in range(V2_RECORD_COUNT):

    record_number = i + 1

    lineage_number = (
        ((record_number - 1) // V2_RECORDS_PER_LINEAGE) + 1
    )

    record_id = f"V2-REC-{record_number:05d}"
    package_id = f"V2-PKG-{lineage_number:04d}"
    lineage_id = f"V2-LINEAGE-{lineage_number:04d}"

    scenario = scenario_assignments[i]

    record = {
        "record_id": record_id,
        "package_id": package_id,
        "lineage_id": lineage_id,
        "scenario_type": scenario,
        "run_id": V2_RUN_ID,
        "record_number": record_number,
        "seed": V2_SEED,
    }

    records.append(record)


# ------------------------------------------------
# Write manifest
# ------------------------------------------------

manifest = {
    "experiment": "Firmware Supply-Chain PoC — Experiment V2",
    "run_id": V2_RUN_ID,
    "seed": V2_SEED,
    "generator_version": V2_GENERATOR_VERSION,
    "record_count": len(records),
    "lineage_count": V2_LINEAGE_COUNT,
    "records_per_lineage": V2_RECORDS_PER_LINEAGE,
    "scenario_distribution": {
        scenario: sum(
            1 for r in records
            if r["scenario_type"] == scenario
        )
        for scenario in SCENARIOS
    },
    "synthetic_data_only": True,
    "generated_at_utc": datetime.now(timezone.utc).isoformat(),
}

manifest_path = HOLDOUT_DIR / "holdout_manifest.json"

manifest_path.write_text(
    json.dumps(manifest, indent=2) + "\n",
    encoding="utf-8",
)


# ------------------------------------------------
# Write records
# ------------------------------------------------

records_path = HOLDOUT_DIR / "holdout_records.csv"

with records_path.open(
    "w",
    newline="",
    encoding="utf-8"
) as f:

    writer = csv.DictWriter(
        f,
        fieldnames=records[0].keys()
    )

    writer.writeheader()
    writer.writerows(records)


# ------------------------------------------------
# Hash the generated files
# ------------------------------------------------

def sha256_file(path):
    h = hashlib.sha256()

    with path.open("rb") as f:
        while chunk := f.read(1024 * 1024):
            h.update(chunk)

    return h.hexdigest()


hashes = {
    "holdout_manifest.json": sha256_file(manifest_path),
    "holdout_records.csv": sha256_file(records_path),
}

(HOLDOUT_DIR / "file_hashes.json").write_text(
    json.dumps(hashes, indent=2) + "\n",
    encoding="utf-8",
)


# ------------------------------------------------
# Validation
# ------------------------------------------------

assert len(records) == V2_RECORD_COUNT
assert len({r["record_id"] for r in records}) == V2_RECORD_COUNT
assert len({r["package_id"] for r in records}) == V2_LINEAGE_COUNT
assert len({r["lineage_id"] for r in records}) == V2_LINEAGE_COUNT

for scenario in SCENARIOS:
    assert sum(
        1 for r in records
        if r["scenario_type"] == scenario
    ) == scenario_per_type


print("=" * 70)
print("V2 INDEPENDENT HOLDOUT GENERATED")
print("=" * 70)

print("Records:", len(records))
print("Lineages:", len({r["lineage_id"] for r in records}))
print("Packages:", len({r["package_id"] for r in records}))
print("Records per lineage:", V2_RECORDS_PER_LINEAGE)

print("\nScenario distribution:")
for scenario, count in manifest["scenario_distribution"].items():
    print(f"  {scenario:25} {count}")

print("\nOutput:")
print(HOLDOUT_DIR)

print("\nManifest:", manifest_path)
print("Records :", records_path)

print("\nGeneration validation: PASS")
print("=" * 70)
```

```python
import pandas as pd
import hashlib
import json

# ------------------------------------------------
# Deterministic helper
# ------------------------------------------------

def stable_hash(value):
    return hashlib.sha256(
        value.encode("utf-8")
    ).hexdigest()


# ------------------------------------------------
# Build observations
# ------------------------------------------------

observations = []

for _, row in v2.iterrows():

    scenario = row["scenario_type"]
    record_id = row["record_id"]

    # Base valid evidence
    observed = {
        "record_id": record_id,
        "package_id": row["package_id"],
        "lineage_id": row["lineage_id"],

        "observed_hash": stable_hash(
            f"trusted-{row['package_id']}"
        ),

        "trusted_hash": stable_hash(
            f"trusted-{row['package_id']}"
        ),

        "signature_valid": True,
        "trusted_root_valid": True,

        "observed_vendor_id": "VEND-001",
        "approved_vendor_id": "VEND-001",

        "observed_version": "2.0.0",
        "approved_version": "2.0.0",

        "rollback_authorized": True,

        "sbom_present": True,
        "sbom_complete": True,

        "vulnerability_present": False,
        "vulnerability_applicable": False,

        "unexpected_component": False,

        "evidence_age_days": 2,
        "freshness_threshold_days": 30,

        "scenario_type": scenario,
    }

    # ------------------------------------------------
    # Scenario-specific observations
    # ------------------------------------------------

    if scenario == "hash_mismatch":
        observed["observed_hash"] = stable_hash(
            f"tampered-{record_id}"
        )

    elif scenario == "invalid_signature":
        observed["signature_valid"] = False

    elif scenario == "untrusted_vendor":
        observed["observed_vendor_id"] = "VEND-004"

    elif scenario == "rollback":
        observed["observed_version"] = "1.0.0"
        observed["approved_version"] = "2.0.0"
        observed["rollback_authorized"] = False

    elif scenario == "missing_sbom":
        observed["sbom_present"] = False
        observed["sbom_complete"] = False

    elif scenario == "vulnerable_component":
        observed["vulnerability_present"] = True
        observed["vulnerability_applicable"] = True

    elif scenario == "unexpected_component":
        observed["unexpected_component"] = True

    elif scenario == "stale_evidence":
        observed["evidence_age_days"] = 90

    elif scenario == "benign_exception":
        # Benign exception remains policy-compliant.
        # It is intentionally not converted into a security failure.
        observed["evidence_age_days"] = 5

    elif scenario == "valid":
        pass

    observations.append(observed)


observations_df = pd.DataFrame(observations)

# ------------------------------------------------
# Required columns
# ------------------------------------------------

required_observation_columns = [
    "record_id",
    "package_id",
    "lineage_id",
    "observed_hash",
    "trusted_hash",
    "signature_valid",
    "trusted_root_valid",
    "observed_vendor_id",
    "approved_vendor_id",
    "observed_version",
    "approved_version",
    "rollback_authorized",
    "sbom_present",
    "sbom_complete",
    "vulnerability_present",
    "vulnerability_applicable",
    "unexpected_component",
    "evidence_age_days",
    "freshness_threshold_days",
    "scenario_type",
]

assert set(required_observation_columns).issubset(
    observations_df.columns
)

assert len(observations_df) == V2_RECORD_COUNT
assert observations_df["record_id"].is_unique

# ------------------------------------------------
# Save
# ------------------------------------------------

observations_path = (
    OBSERVED_DIR / "v2_observations.csv"
)

observations_df.to_csv(
    observations_path,
    index=False
)

# ------------------------------------------------
# Audit
# ------------------------------------------------

print("=" * 70)
print("V2 OBSERVATIONS GENERATED")
print("=" * 70)

print("Rows:", len(observations_df))
print("Columns:", len(observations_df.columns))

print("\nScenario counts:")
print(
    observations_df["scenario_type"]
    .value_counts()
    .sort_index()
    .to_string()
)

print("\nObservation file:")
print(observations_path)

print("\nObservation generation: PASS")
print("=" * 70)
```

```python
RISK_WEIGHTS = {
    "hash_integrity_failure": 40,
    "signature_failure": 30,
    "trusted_root_failure": 20,
    "untrusted_vendor": 40,
    "unauthorized_rollback": 40,
    "vulnerable_component": 20,
    "unexpected_component": 20,
    "missing_sbom": 15,
    "stale_evidence": 10,
}


def version_tuple(value):
    return tuple(int(x) for x in str(value).split("."))


def evaluate_policy(row):
    factors = []

    if row["observed_hash"] != row["trusted_hash"]:
        factors.append("hash_integrity_failure")

    if not bool(row["signature_valid"]):
        factors.append("signature_failure")

    if not bool(row["trusted_root_valid"]):
        factors.append("trusted_root_failure")

    if (
        row["observed_vendor_id"] != row["approved_vendor_id"]
        or row["observed_vendor_id"] == "VEND-004"
    ):
        factors.append("untrusted_vendor")

    if (
        version_tuple(row["observed_version"])
        < version_tuple(row["approved_version"])
        and not bool(row["rollback_authorized"])
    ):
        factors.append("unauthorized_rollback")

    if (
        not bool(row["sbom_present"])
        or not bool(row["sbom_complete"])
    ):
        factors.append("missing_sbom")

    if (
        bool(row["vulnerability_present"])
        and bool(row["vulnerability_applicable"])
    ):
        factors.append("vulnerable_component")

    if bool(row["unexpected_component"]):
        factors.append("unexpected_component")

    if (
        float(row["evidence_age_days"])
        > float(row["freshness_threshold_days"])
    ):
        factors.append("stale_evidence")

    score = min(
        100,
        sum(RISK_WEIGHTS[factor] for factor in factors)
    )

    if score >= 40:
        label = "high_risk"
    elif score > 0:
        label = "investigate"
    else:
        label = "normal"

    return score, label, factors


# ------------------------------------------------
# Apply policy
# ------------------------------------------------

policy_results = []

for _, row in observations_df.iterrows():

    score, label, factors = evaluate_policy(row)

    policy_results.append({
        "record_id": row["record_id"],
        "scenario_type": row["scenario_type"],
        "risk_score": score,
        "ground_truth_label": label,
        "risk_factors": json.dumps(factors),
    })


policy_df = pd.DataFrame(policy_results)


# ------------------------------------------------
# Validation
# ------------------------------------------------

assert len(policy_df) == V2_RECORD_COUNT
assert policy_df["record_id"].is_unique

assert set(
    policy_df["ground_truth_label"]
) <= {
    "normal",
    "investigate",
    "high_risk",
}


# ------------------------------------------------
# Save ground truth
# ------------------------------------------------

ground_truth_path = (
    LABEL_DIR / "v2_ground_truth.csv"
)

policy_df.to_csv(
    ground_truth_path,
    index=False
)


# ------------------------------------------------
# Compact audit
# ------------------------------------------------

print("=" * 70)
print("V2 DETERMINISTIC POLICY EVALUATION")
print("=" * 70)

print("Records:", len(policy_df))

print("\nGround-truth distribution:")
print(
    policy_df["ground_truth_label"]
    .value_counts()
    .sort_index()
    .to_string()
)

print("\nGround truth by scenario:")
print(
    policy_df
    .groupby(
        ["scenario_type", "ground_truth_label"]
    )
    .size()
    .unstack(fill_value=0)
    .to_string()
)

print("\nGround-truth file:")
print(ground_truth_path)

print("\nPolicy evaluation: PASS")
print("=" * 70)
```
